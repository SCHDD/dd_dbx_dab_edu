```python
# define var
file_path = '/Volumes/training/sch/study_volume/games_ranking.csv'
table = 'training.sch.bronze_games_ranking'
```

```python
# define schema
schema = '''
    game_name STRING,
    genre STRING,
    rank_type STRING,
    rank STRING
'''
```

```python
# read dataframe read & write table

df = spark.read.schema(schema).csv(file_path, header=True)

# for skip exists table
df.write.mode('overwrite').saveAsTable(table)
```
